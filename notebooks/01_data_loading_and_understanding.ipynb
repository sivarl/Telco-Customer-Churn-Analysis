#### Key Observations

 - The dataset contains **7043 customer records** with **33 columns**.
 - There are **32 feature variables** and **1 target variable**.

    - 0 indicates customer retention
      
 - The dataset containes **9 numerical features** (3 float, 6 integer) and **24 categorical features**.
 
 - Most columns do not contain missing values.
 - The **'Churn Reason'** has **5174 missing values**, which likely correspond to customers who do not churn.
 - The missing values in **'Churn Reason'** are expected, since churn reasons are only recorded for customers who have churned.
